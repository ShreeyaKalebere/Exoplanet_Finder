Run `python -m src.get_tce_table` and `python -m src.download_lightcurves --max-stars 300` first.

```python
import sys; sys.path.insert(0, '..')
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from src.config import TCE_CSV, KEPLER90_KEPID
from src.views import load_star, flatten, make_views

tce = pd.read_csv(TCE_CSV)
tce.av_training_set.value_counts()
```

```python
# raw vs flattened light curve of Kepler-90
star = tce[tce.kepid == KEPLER90_KEPID]
lc = load_star(KEPLER90_KEPID)
t, f = flatten(lc, star.tce_period, star.tce_time0bk, star.tce_duration)
fig, ax = plt.subplots(2, 1, figsize=(12, 5), sharex=True)
ax[0].plot(lc.time.value, lc.flux.value, ',', color='k'); ax[0].set_title('raw (stitched)')
ax[1].plot(t, f, ',', color='k'); ax[1].set_title('flattened, transits masked during the fit')
plt.show()
```

```python
# global and local view of each Kepler-90 TCE
for r in star.sort_values('tce_period').itertuples():
    g, l = make_views(t, f, r.tce_period, r.tce_time0bk, r.tce_duration)
    fig, ax = plt.subplots(1, 2, figsize=(12, 2.5))
    ax[0].plot(g, '.', ms=1); ax[0].set_title(f'{r.tce_id} ({r.av_training_set}) global, P={r.tce_period:.2f} d')
    ax[1].plot(l, '.-', ms=2); ax[1].set_title('local')
    plt.show()
```
